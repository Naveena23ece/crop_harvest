Runs a mini-training session on a small data subset to verify the pipeline works end-to-end.

```python
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), '..'))

import torch
import numpy as np
import matplotlib.pyplot as plt

from training.model_artcn import build_artcn, ARTCN
print(f'PyTorch version: {torch.__version__}')
print(f'CUDA available: {torch.cuda.is_available()}')
```

```python
# Build the AR-TCN model and inspect architecture
model = build_artcn(n_features=60, n_classes=3)
print(model)
print(f'\nTotal trainable parameters: {model.count_parameters():,}')
```

```python
# Test forward pass with random data
batch = torch.randn(8, 24, 60)   # (batch=8, seq=24, features=60)
model.eval()
with torch.no_grad():
    logits = model(batch)
    proba  = torch.softmax(logits, dim=-1)

print(f'Input:   {batch.shape}')
print(f'Logits:  {logits.shape}')
print(f'Proba:   {proba.shape}')
print(f'Sample output (first item):')
for i, cls in enumerate(['Low', 'Moderate', 'High']):
    print(f'  {cls}: {proba[0, i]:.3f}')
```
