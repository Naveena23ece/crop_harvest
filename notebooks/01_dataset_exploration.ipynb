This notebook explores the generated synthetic dataset and validates its statistical properties.

```python
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), '..'))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from dataset.generate_dataset import generate_dataset

plt.style.use('dark_background')
plt.rcParams['figure.figsize'] = (12, 5)
print('Libraries loaded.')
```

```python
# Generate a small sample for exploration
df = generate_dataset(20_000)
print(df.shape)
df.head()
```

```python
# Statistical summary
df.describe().round(2)
```

```python
# Class distribution
fig, ax = plt.subplots()
vc = df['risk_category'].value_counts()
colors = ['#22c55e', '#f59e0b', '#ef4444']
ax.bar(vc.index, vc.values, color=colors)
ax.set_title('Risk Category Distribution')
ax.set_ylabel('Count')
for i, (idx, val) in enumerate(vc.items()):
    ax.text(i, val + 50, f'{val:,}\n({val/len(df)*100:.1f}%)', ha='center', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
numeric_cols = ['temperature_c', 'humidity_pct', 'gas_concentration',
                'grain_moisture', 'storage_duration', 'risk_score']
fig, ax = plt.subplots(figsize=(8, 6))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdYlGn',
            ax=ax, center=0, linewidths=0.5)
ax.set_title('Feature Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
# Time series: first 500 readings
sample = df.head(500)
fig, axes = plt.subplots(4, 1, figsize=(14, 10), sharex=True)
for ax, col, color in zip(axes,
    ['temperature_c', 'humidity_pct', 'gas_concentration', 'grain_moisture'],
    ['#f97316', '#3b82f6', '#a855f7', '#eab308']):
    ax.plot(sample.index, sample[col], color=color, lw=1)
    ax.set_ylabel(col.replace('_', ' '))
    ax.grid(True, alpha=0.2)
axes[0].set_title('Sensor Time Series (first 500 readings)')
plt.tight_layout()
plt.show()
```

```python
# Missing value analysis
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({'count': missing, 'pct': missing_pct})[missing > 0]
```
